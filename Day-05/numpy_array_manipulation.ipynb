```python
# Day 5: NumPy Array Manipulation

import numpy as np
```

```python
# Program 1: Reshape an Array

arr = np.arange(1, 13)

print("Original Array:")
print(arr)

print("\nReshaped Array:")
print(arr.reshape(3, 4))
```

```python
# Program 2 : Flatten a 2D Array

arr = np.array([
    [1,2,3] ,
    [4,5,6]
])

print(arr.flatten())
```

```python
# Program 3 : Ravel on Array

arr = np.array([
    [1,2,3] ,
    [4,5,6]
])

print(arr.ravel())
```

```python
# Program: Difference Between ravel() and flatten()

import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

# Create ravel and flatten arrays
ravel_arr = arr.ravel()
flatten_arr = arr.flatten()

# Modify both arrays
ravel_arr[0] = 100
flatten_arr[1] = 200

print("Original Array:")
print(arr)

print("\nRavel Array:")
print(ravel_arr)

print("\nFlatten Array:")
print(flatten_arr)

# ravel() : returns a view, so changing ravel_arr also changes the original array.
# flatten() : returns a copy, so changing flatten_arr does not affect the original array
```

```python
# Program 4: Transpose a Matrix

arr = np.array([[1,2,3],
                [4,5,6]])

print(arr.T)
```

```python
# Program 5 : Resize an Array

arr = np.array([1,2,3,4,5,6])

print(np.resize(arr,(3,4)))
```

```python
# Program 6 : Change Shape

arr = np.arange(1,17)

print(arr.shape)

new_arr = arr.reshape(4,4)

print(new_arr)
```

```python
# Program 7 : Flatten Using reshape()

arr = np.array([
    [10 , 20] ,
    [30,40]
])

print(arr.reshape(-1))
```

```python
# Program 8 : Convert 1D to 2D

arr = np.arange(1,10)

print(arr.reshape(3,3))
```

```python
# Program 9 : Convert 2D to 1D

arr = np.array([
    [1 , 2 , 3] ,
    [4 , 5 , 6]  
])

print(arr.flatten())
```

```python
# Program 10 : Copy vs View

arr = np.array([10 , 20 , 30 , 40 ])

copy_arr =  arr.copy()
view_arr = arr.view()

arr[0] = 100

print("Original : ",arr)

print("Copy : ",copy_arr)

print("View : ",view_arr)
```

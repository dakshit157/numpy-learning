```python
# Day 2 - NumPy Array Creation Functions

import numpy as np
```

```python
# Program 1: Create a 1D array of zeros

arr1 = np.zeros(6)

print(arr1)
```

```python
# Program 2: Create a 2D array of zeros

arr2 = np.zeros((3,3))

print(arr2)
```

```python
# Program 3: Create a 1D array of ones

arr3 = np.ones(6)

print(arr3)
```

```python
# Program 4: Create a 2D array of ones

arr = np.ones((2,3))
print(arr)
```

```python
# Program 5: Create an array using arange()

arr = np.arange(6)

print(arr)
```

```python
# Program 6: Create an array using arange() with step

arr = np.arange(1,20,2)

print(arr)
```

```python
# Program 7: Create an array using linspace()

arr = np.linspace(1,10,5)

print(arr)
```

```python
# Program 8: Create an identity matrix

arr = np.eye(3)

print(arr)
```

```python
# Program 9: Generate random decimal numbers

# Random decimal numbers
arr = np.random.rand(5)
print(arr)

# Random integers between 1 and 10
arr1 = np.random.randint(1 , 10 , 5)
print(arr1)
```

```python
# Mini Project: Student Roll Numbers

roll_numbers = np.arange(101,111)

print("Student Roll Numbers:")
print(roll_numbers)

identity = np.eye(5)
print("\nIdentity Matrix:")
print(identity)

random_marks = np.random.randint(50,101,10)
print("\nRandom Marks:")
print(random_marks)
```

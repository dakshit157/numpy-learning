```python
import numpy as np
```

```python
# Program 1: Create a 1D Array

arr1 = np.array([ 1 , 2 , 3 , 4 , 5 ])
print(arr1)
```

```python
# Program 2: Create a 2D Array

arr2 = np.array([[1 , 2 , 3 ] ,
                 [ 4 , 5 , 6 ]])

print(arr2)
```

```python
# Program 3: Find Shape

print(arr2.shape)
```

```python
# Program 4: Find Dimensions

print(arr2.ndim)
```

```python
# Program 5: Find Size

print(arr2.size)
```

```python
# Program 6: Find Data Type

print(arr1.dtype)
print(arr2.dtype)
```

```python
# Program 7: Convert List to NumPy Array

numbers = [ 1 , 2 , 3 , 4 , 5 , 6 ]

arr = np.array(numbers)

print(arr)

```

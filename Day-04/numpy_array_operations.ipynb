```python
# Day 4: NumPy Array Operations

import numpy as np
```

```python
# Program 1 : Addition of Two Arrays

arr1 = np.array([10 , 20 , 30])
arr2 = np.array([5 , 10 , 15])

print(arr1 + arr2)
```

```python
# Program 2 : Substraction of Two Arrays

arr1 = np.array([10 , 20 , 30])
arr2 = np.array([5 , 10 , 15])

print(arr1 - arr2)
```

```python
# Program 3 : Multiplication of Two Arrays 

arr1 = np.array([2, 4, 6])
arr2 = np.array([3, 5, 7])

print(arr1 * arr2)
```

```python
# Program 4: Division of Two Arrays

arr1 = np.array([20, 40, 60])
arr2 = np.array([2, 4, 5])

print(arr1 / arr2)
```

```python
# Program 6: Square Root
 
arr = np.array([2 , 3 , 4])

print(arr ** 2)
```

```python
arr = np.array([2 , 3 , 4])

print(np.sum(arr))
```

```python
# Program 8: Mean, Maximum , Minimum and Standard Deviation

arr = np.array([10,20,30,40,50])

print("\nMean : ",np.mean(arr))

print("\nMaximum : ",np.max(arr))

print("\nMinimum : ",np.min(arr))

print("\nStandard Deviation : ",np.std(arr))
```

```python
# Program 9 : Reshape an Array

arr = np.arange(1 , 10)

print(arr.reshape(3,3))
```

```python
# Mini Project: Student Marks Analysis

marks = np.array([75, 82, 91, 68, 88])

print("Marks:", marks)

print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Standard Deviation:", np.std(marks))

print("\nReshaped Array:")
print(marks.reshape(5,1))
```

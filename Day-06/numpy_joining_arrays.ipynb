```python
# Day 6: NumPy Joining Arrays

import numpy as np
```

```python
# Program 1 : Concatenate 1D Arrays

arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

result = np.concatenate((arr1,arr2))

print(result)
```

```python
# Program 2 : Concatenate 2D Arrays

arr1 = np.array([[1, 2],
                 [3, 4]])

arr2 = np.array([[5, 6],
                 [7, 8]])



# For rows

print(np.concatenate((arr1,arr2) , axis = 0))

print("\n")

# For columns

print(np.concatenate((arr1 , arr2) , axis = 1))

```

```python
# Program 3 : Horizontal Stack (hstack)

arr1 = np.array([10 , 20 , 30])
arr2 = np.array([40 , 50 , 60])

result = np.hstack((arr1,arr2))

print(result)
```

```python
# Program 4 : Vertical Stack (vstack)

arr1 = np.array([10 , 20 , 30])
arr2 = np.array([40 , 50 , 60])

result = np.vstack((arr1,arr2))

print(result)
```

```python
# Program 5 : Depth Stack (dstack)

arr1 = np.array([10 , 20 , 30])
arr2 = np.array([40 , 50 , 60])

result = np.dstack((arr1,arr2))

print(result)
```

```python
# Program 6 : Column Stack

arr1 = np.array([10 , 20 , 30])
arr2 = np.array([40 , 50 , 60])

result = np.column_stack((arr1,arr2))

print(result)
```

```python
# Program 7 : Stack() 

arr1 = np.array([10 , 20 , 30])
arr2 = np.array([40 , 50 , 60])

print("Axis = 0")
print(np.stack((arr1,arr2) , axis=0))

print("\nAxis = 1")
print(np.stack((arr1,arr2), axis=1))
```

```python
## Mini Project: Student Data Joining

roll_no = np.array([101, 102, 103])
marks = np.array([85, 90, 78])

print("Column Stack:")
print(np.column_stack((roll_no, marks)))

print("\nHorizontal Stack:")
print(np.hstack((roll_no, marks)))

print("\nVertical Stack:")
print(np.vstack((roll_no, marks)))
```

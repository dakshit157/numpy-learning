```python
# Day 9 : NumPy Sorting and Searching

import numpy as np
```

```python
# Program 1 : Sort a 1D Array

arr = np.array([5,7,10,1,2,9])

sorted = np.sort(arr)

print(sorted)
```

```python
# Program 2 : Sort a String Array

arr = np.array(["Banana" , "Apple" , "Mango" , "Orange"])

sorted = np.sort(arr)

print(sorted)
```

```python
# Program 3 : Sort a 2D Array

arr = np.array([
    [1 , 8 , 7],
    [3 , 10 , 9]
])

sorted = np.sort(arr)

print(sorted)
```

```python
# Program 4 : argsort()

arr = np.array([50 , 20 , 10 , 40 , 30])

print("Original Array : ",arr)

print("Sorted Indexes : ")
print(np.argsort(arr))
```

```python
# Program 5 : where()

arr = np.array([50 , 20 , 10 , 20 , 30 , 20])

print(np.where(arr == 20))
```

```python
# Program 6 : searchsorted()

arr = np.array([10,20,30,40,50])

print(np.searchsorted(arr , 25))
```

```python
# Program 7: searchsorted() Multiple Values

arr = np.array([10, 20, 30, 40, 50])

print(np.searchsorted(arr, [15, 25, 45]))
```

```python
# Program 8 : Extract Even Numbers

arr = np.array([2 , 5 , 7 , 4 , 11 , 12 , 14 , 16 , 15])

print(arr[arr % 2 ==0])
```

```python
# Program 9 : Extract Odd Numbers

arr = np.array([2 , 5 , 7 , 4 , 11 , 12 , 14 , 16 , 15])

print(arr[arr % 2 != 0])
```

```python
# Program 10 : Find Maximum and Minimum Index

arr = np.array([2 , 5 , 7 , 4 , 11 , 12 , 14 , 16 , 15])

print("Maximum Index : ",np.argmax(arr))

print("Minimum Index : ",np.argmin(arr))
```

```python
## Mini Project: Student Marks Analysis

marks = np.array([75, 92, 68, 88, 95, 81])

print("Original Marks:")
print(marks)

print("\nSorted Marks:")
print(np.sort(marks))

print("\nHighest Marks Index:")
print(np.argmax(marks))

print("\nLowest Marks Index:")
print(np.argmin(marks))

print("\nMarks Greater Than 80:")
print(marks[marks > 80])
```
